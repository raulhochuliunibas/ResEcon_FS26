Using another package `pandas` you can also group multiple arrays togehter to have a data frame which is easier to handle than a bunch of loose arrays. 

```python
# Data Frames 
import pandas as pd
data = {
    "Year": [2020, 2021, 2022, 2023],
    "Price": [50, 55, 60, 65],
    "Quantity": [100, 98, 95, 93]
}

df = pd.DataFrame(data)
df
```

You can use data frames to filter data by one or more columsn and also add new columns to make life easier for your computations. 

```python
# Basic operations + filering
df['Price'] # call Price column
df.iloc[0]  # call first row 

# filtering
df.loc[df['Year'] == 2022]
df.loc[df['Price'] > 55]

# column operations
df["Revenue"] = df["Price"] * df["Quantity"]
df
```

## Functions

A function is a small, reusable block of code that performs a specific task. This is handy, if a specific operation is not available in a existing command or a package. Instead of writing the same code again and again, we define it once and then call it whenever we need it.

```python
# A function that adds two numbers

def add_numbers_round(x, y):
    result = x + y
    result = round(result,1)
    return result

print(4.12354 + 10)
print(add_numbers_round(4.12354, 10))

print(9.7 + 8.241534)
print(add_numbers_round(9.7, 8.241534))
```

Functions can have default values, that don't need to be specified. 

```python
def add_numbers_round(x, y, round_digits=2):
    result = x + y
    result = round(result, round_digits)
    return result

print(4.12354 + 10)
print(add_numbers_round(4.12354, 10))
print(add_numbers_round(4.12354, 10,  4))

print(9.7 + 8.241534)
print(add_numbers_round(9.7, 8.241534))
print(add_numbers_round(9.7, 8.241534, 6))
```

Instead of default values, you can also use pre-defined variables within the function. But be careful about naming conventions and unwanted dependencies. Values outside the `return`statement are **not stored at all**.

```python
discount_rate = 0.125
def discounted_price_1(price):
    discounted = price * (1 - discount_rate)
    return discounted
discounted_price_1(100)
```

```python
discount_rate = 0.125
def discounted_price_2(price, discount_rate=0.1):
    # function variables override the naming convention for global variables
    discounted = price * (1 - discount_rate)
    return discounted
print(f'discounted_price_1: {discounted_price_1(100)}')
print(f'discounted_price_2: {discounted_price_2(100)}')
print(f'discounted_price_2 =/= discounted_price_1, despite the global variable discount_rate remains: {discount_rate}')
```

```python
def compute_cost(price, quantity):
    total_cost = price * quantity
    add_overhead_cost = total_cost * 0.25  # not returned
    return total_cost + add_overhead_cost
total = compute_cost(50, 10)
add_overhead_cost
```
